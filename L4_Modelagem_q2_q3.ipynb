```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import odeint
```

### LETRA D

```python
# Define o valor global de u
u = -0.429505

# Define as condições iniciais
initial_conditions = [
    np.array([-0.3, 0.3]),
    np.array([-0.3, 0.3]),
    np.array([0.01, 0]),
    np.array([-1.8, 3.9]),
    np.array([0.7, 0])
]

# Define as equações diferenciais
def bifHopf(x, t):
    dxdt = np.zeros(2)
    dxdt[0] = x[1]
    dxdt[1] = x[0] - x[0]**2 + u * x[1] + 0.5 * x[0] * x[1]
    return dxdt

# Define os intervalos de tempo para integração
tspan = np.linspace(0, 20, 1000)
```

```python
# Lista de rótulos para os gráficos
labels = ['Chute 1', 'Chute 2', 'Chute 3', 'Chute 4', 'Chute 5']

# Configura a plotagem
plt.figure(figsize=(8, 8))
plt.title(f'Plano de Fases para u = {u}')
plt.xlabel('X_1')
plt.ylabel('X_2')

# Resolve as equações diferenciais e plota as órbitas
for i, ic in enumerate(initial_conditions):
    xt = odeint(bifHopf, ic, tspan)
    plt.plot(xt[:, 0], xt[:, 1], label=labels[i], linewidth=1)

plt.legend()
plt.grid()
plt.xlim(-6, 6)
plt.ylim(-6, 6)
plt.show()
```

### LETRA E

```python
# Definir as equações diferenciais
def sistema(x, t, mu):
    x1, x2 = x
    dx1dt = x2
    dx2dt = x1 - x1**2 + mu*x2 + 0.5*x1*x2
    return [dx1dt, dx2dt]

# Parâmetro mu
mu = -0.45

# Condição inicial para a simulação a)
x0 = [0, 0.3]
t = np.linspace(0, 12, 1000)  # Tempo de 0 a 12, com 1000 pontos
sol_a = odeint(sistema, x0, t, args=(mu,))

# Condição inicial para a simulação b)
x0_b = [0, 0.299]  # Um pouco menor que 0.3
sol_b = odeint(sistema, x0_b, t, args=(mu,))

# Condição inicial para a simulação c)
x0_c = [0.001, 0.0008]
sol_c = odeint(sistema, x0_c, t, args=(mu,))
```

```python
# Plotar as trajetórias
plt.figure(figsize=(12, 4))

# a) Trajetória para mu = -0.45, partindo de (0, 0.3)
plt.subplot(131)
plt.plot(sol_a[:, 0], sol_a[:, 1])
plt.title('a) Trajetória para mu = -0.45, (0, 0.3)')
plt.xlabel('x1')
plt.ylabel('x2')

# b) Trajetória para mu = -0.45, partindo de (0, 0.299)
plt.subplot(132)
plt.plot(sol_b[:, 0], sol_b[:, 1])
plt.title('b) Trajetória para mu = -0.45, (0, 0.299)')
plt.xlabel('x1')
plt.ylabel('x2')

# c) Trajetória para mu = -0.45, partindo de (0.001, 0.0008)
plt.subplot(133)
plt.plot(sol_c[:, 0], sol_c[:, 1])
plt.title('c) Trajetória para mu = -0.45, (0.001, 0.0008)')
plt.xlabel('x1')
plt.ylabel('x2')

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import odeint
```

### Figura 17

```python
# Define the global variables
Da = 0.20
B = 16.2
beta = 3
x2c = 0

# Define the differential equations
def figura_17(x, t):
    dxdt = np.zeros(2)
    dxdt[0] = -x[0] + Da * (1 - x[0]) * np.exp(x[1])
    dxdt[1] = -x[1] + Da * B * (1 - x[0]) * np.exp(x[1]) - beta * (x[1] - x2c)
    return dxdt

# Initial conditions and time span
tspan = np.linspace(0, 15, 1000)
x01 = np.array([0.9, 2.1])
x02 = np.array([0.95, 3.3])

# Solve the ODEs
xt1 = odeint(figura_17, x01, tspan)
xt2 = odeint(figura_17, x02, tspan)

# Plot the results
plt.figure()
plt.plot(xt1[:, 0], xt1[:, 1], color='red', linewidth=2)
plt.plot(xt2[:, 0], xt2[:, 1], color='orange', linewidth=2)
plt.scatter(x01[0], x01[1], color='red', label='Ponto inicial 1', marker='x')
plt.scatter(x02[0], x02[1], color='orange', label='Ponto inicial 2', marker='x')
plt.title('Fig. 17 - Plano de Fases')
plt.xlabel('X1')
plt.ylabel('X2')
plt.axis([0, 1.7, 0, 12])
plt.legend(['Sim. Dinâmica 1', 'Sim. Dinâmica 2', 'Ponto inicial 1', 'Ponto inicial 2'])
plt.grid(True)
plt.show()
```
